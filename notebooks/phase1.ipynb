## ۱. منشأ، واحد و محدودیت تفسیر

- ورودی‌ها زیرمجموعهٔ ساده‌شده‌ای از NACC هستند؛ تاریخ دریافت به نقل از دانشجو ۱۰ اکتبر است و سال/دورهٔ استخراج در اطلاعات تحویلی تصریح نشده است.

## ۲. محیط اجرا و قواعد ثابت

اجرای ممیزی، ساخت جدول و خلاصه‌های توصیفی با کتابخانهٔ استاندارد Python است. رسم شکل‌ها به NumPy و Matplotlib طبق requirements.txt نیاز دارد؛ با نسخه‌های قفل‌شدهٔ فعلی، محیط Python 3.11 یا جدیدتر لازم است. برای اجرای این نوت‌بوک، محیط سازگار با ipykernel لازم است. قواعد عددی همراه منابع در پروژه موجودند و در زمان اجرا از اینترنت خوانده نمی‌شوند.

از پوشهٔ ریشه یا notebooks همهٔ سلول‌ها را به ترتیب اجرا کنید. این نوت‌بوک گزارش‌های ممیزی، جدول مرجع مشتق همراه جریان نمونه و جدول‌ها و گزارش آمار توصیفی و چهار شکل PNG/SVG را بازتولید می‌کند؛ CSVهای ورودی را بازنویسی نمی‌کند. جدول مرجع همهٔ افراد مشترک را نگه می‌دارد و نمونهٔ هر خروجی با پرچم آن انتخاب می‌شود.

اجرای کامل بدون نصب هستهٔ عمومی Jupyter، از ریشهٔ پروژه:

```bash
python3 -m pip install -r requirements.txt
python3 scripts/run_phase1_notebook.py
```

این فرمان با همان Python یک هستهٔ تازه می‌سازد، سلول‌ها را به ترتیب اجرا می‌کند و پس از موفقیت خروجی‌ها را در همین نوت‌بوک نگه می‌دارد. برای مشاهدهٔ خروجی موجود اجرای مجدد لازم نیست. بخش‌های محاسباتی به اینترنت نیاز ندارند؛ نصب اولیهٔ بسته‌ها ممکن است دسترسی به منبع بسته‌ها بخواهد. فایل‌های منبع فقط خوانده می‌شوند؛ گزارش‌ها و خروجی‌های مشتق در مسیرهای مشخص بازتولید می‌شوند.

</div>


````python
from pathlib import Path
import json
import re
import sys
from collections import Counter
from IPython.display import HTML, Markdown, display

candidates = [Path.cwd(), *Path.cwd().parents]
ROOT = next((p for p in candidates if (p / "config/phase1_quality_rules.json").is_file() and (p / "data/udslong_imputed.csv").is_file()), None)
if ROOT is None:
    raise FileNotFoundError("نوت‌بوک را از ریشهٔ پروژه یا پوشهٔ notebooks اجرا کنید.")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
from src.phase1_quality import audit, write_reports
from scripts.build_phase1_pdf import notebook_font_css
display(HTML(notebook_font_css(ROOT)))

def show_table(headers, rows):
    def cell(v):
        return str(v).replace("|", chr(92) + "|").replace("\n", " ")
    lines = ["| " + " | ".join(map(cell, headers)) + " |", "| " + " | ".join(["---"] * len(headers)) + " |"]
    lines += ["| " + " | ".join(map(cell, row)) + " |" for row in rows]
    display(Markdown("\n".join(lines)))

def display_section(text):
    nested = re.sub(r"^(#{2,5}) ", lambda match: "#" + match.group(1) + " ", text, flags=re.MULTILINE)
    display(Markdown(nested))

config = json.loads((ROOT / "config/phase1_quality_rules.json").read_text(encoding="utf-8"))
assert config["moca_fractional_policy"] == "retain_in_range_flag_and_integer_sensitivity"
print("قواعد بارگذاری شد؛ سیاست اعشار MoCA مطابق انتخاب دانشجو است.")
````

## ۱۳. صورت نهایی سؤال، مناسبت داده و جمعیت مطالعه

**تصمیم نهایی سؤال: سؤال و شاخص انتخاب‌شده حفظ می‌شوند؛ تغییر موضوع یا انتخاب ویژگی بر اساس نتیجه انجام نشده است.** وجود پاسخ و اندازه‌گیری MRI در سطح همان شناسه، تعریف روشن شاخص و تعداد قابل استفاده، ادامهٔ بررسی ارتباطی را از نظر ساختار داده ممکن می‌کند. این امکان، اعتبار بدون قید اندازه‌گیری‌ها یا تأیید فرضیهٔ علمی نیست؛ محدودیت‌های زیر در تمام فازها همراه نتیجه می‌مانند.

عنوان تثبیت‌شده برای فاز یک:

**ارتباط میانگین ضخامت قشر انتورینال دو نیمکره با نمرهٔ MoCA در زیرمجموعهٔ تحویلی NACC**

> در زیرمجموعهٔ تحویلی NACC، میانگین ضخامت قشر انتورینال چپ و راست چه رابطه‌ای با نمرهٔ MoCA دارد و این رابطه با در نظر گرفتن سن، جنس و سال‌های تحصیل چگونه است؟

منظور از MoCA در اجرای این پروژه، **نمرهٔ کاریِ آماده‌شدهٔ موجود در فایل تحویلی** است: مقادیر داخل دامنه طبق سیاست مصوب، همراه پرچم اعشار. این بیان به نمرهٔ اولیهٔ قبل از جایگذاری، نمرهٔ اصیل تأییدشده، اصلاح آموزشی تازه یا توان شناختی با مقیاس نسبت تبدیل نمی‌شود. عنوان، سؤال، دو ورودی شاخص، وزن مساوی و متغیرهای همراه از روی نمودارها یا اختلاف گروه‌ها عوض نشده‌اند.

#### تعریف عملیاتی و واحد مطالعه

- پاسخ: `MOCATOTS_ANALYSIS` از UDS؛ نمرهٔ کاری در دامنهٔ ۰ تا ۳۰، بدون گردکردن یا برش نمره.
- ویژگی MRI: `ENTORHINAL_THICKNESS_MEAN = (LH_ENTORHINAL_AVGTH + RH_ENTORHINAL_AVGTH) / 2`، به شرط کفایت هر دو ورودی همان فرد؛ واحد mm. وزن‌دهی با سطح قشر و جایگزینی یک نیمکره به‌تنهایی تعریف نشده است.
- متغیرهای همراه: سن ثبت‌شده، جنس و سال‌های تحصیل. گروه `NACCUDSD` برای توصیف گروه‌هاست و وارد فهرست پیش‌بین‌های اصلی نشده است.
- واحد تحلیل: فرد با شناسهٔ `NACCID`؛ در همین فایل‌ها یک رکورد به‌ازای هر شناسه وجود دارد. این وضعیتِ تحویل با تک‌مراجعه‌ای‌بودن منبع یا استقلال همهٔ افراد/خوشه‌ها معادل نیست. ساختار منبع طولی یا قاعدهٔ انتخاب مراجعه از فایل فعلی بازسازی نمی‌شود.
- دامنهٔ تفسیر: رکوردهای قابل استفادهٔ همین زیرمجموعهٔ آموزشی NACC و هر خروجی با مخرج خودش. این مطالعه تغییر طولی، علت بیماری، تشخیص اختصاصی آلزایمر یا اثر درمان را برآورد نمی‌کند.

## ۱۶. ادامهٔ تحلیل و تصمیم‌های ثابت

در فازهای بعد، با پاسخ MoCA و افراد یکسان، مدل UDS-only شامل سن، جنس و تحصیلات، مدل MRI-only شامل میانگین ضخامت انتورینال و مدل ترکیبی مقایسه خواهند شد. مجموعهٔ مشترک اصلی فعلاً ۱۲۲۶ و مجموعهٔ مشترک حساسیت ۹۸۲ فرد دارد؛ مدل‌ها هنوز برازش نشده‌اند.

نوع آزمون، کدگذاری متغیرها و بررسی شکل رابطه و باقیمانده‌ها در فاز مربوط مستند می‌شود. سناریوی صحیح‌مقدار، دادهٔ پیش از جایگذاری را بازسازی نمی‌کند. پرچم IQR قاعدهٔ حذف نیست؛ اثرگذاری مشاهده‌ها در مدل موضوع جداگانه‌ای است.

قاعدهٔ ورود برای ادامه، حداکثر یک رکورد تحویلی به‌ازای هر NACCID است. چون تاریخ و شمارهٔ مراجعه موجود نیست، انتخاب اولین یا آخرین مراجعه ادعا نمی‌شود. برای بازسازی تاریخچهٔ واقعی مراجعه یا بررسی تطبیق زمانی به فرادادهٔ تیم درس نیاز است.

## ۱۷. منابع و راهنمای بازتولید

منابع زیر پشتوانهٔ تعریف‌ها، قواعد و شمارهای گزارش‌اند. نسخهٔ مشخص یا آماده‌سازی زیرمجموعه از مستندات عمومی تعیین نشده است. شمارهٔ صفحه در پیوند PDF از ابتدای فایل حساب می‌شود.

- [داک درس، فاز یک](../docs/amarpr.pdf#page=6)، صفحات ۶ تا ۸: شناخت داده، توصیف، مقایسه و نهایی‌کردن سؤال؛ [مسیر ج در فاز چهارم](../docs/amarpr.pdf#page=15): پاسخ کمی، سه مدل و محدودیت مراجعه/تطبیق زمانی.
- [کارت پروژه](../PROJECT_CARD.md)، [فرهنگ متغیرهای منتخب](../docs/SELECTED_VARIABLES.md) و [قواعد کیفیت](../docs/QUALITY_RULES.md): تعریف‌ها، پیوند منابع رسمی و سیاست‌های منتخب. تعریف رسمی متغیر، نسخه و آماده‌سازی فایل را اثبات نمی‌کند.
- [ممیزی](../reports/phase1/quality_audit.md)، [جریان نمونه](../reports/phase1/sample_flow.md)، [گزارش توصیفی](../reports/phase1/descriptive_statistics.md)، [نتایج توصیفی JSON](../reports/phase1/descriptive_statistics.json) و [چهار شکل](../reports/phase1/figures.md): تمام شمارها و یافته‌های گزارش‌شده از همین فایل‌هاست.
- [شناسنامهٔ جدول مرجع](../data/processed/phase1_manifest.json) و [متادیتای شکل‌ها](../reports/phase1/figures_metadata.json): تعریف مجموعه‌ها و ردگیری منبع/کد/خروجی؛ [نوت‌بوک](phase1.ipynb) خروجی‌ها و همین جمع‌بندی را کنار هم نگه می‌دارد.
### منابع رسمی تعریف متغیرها

- [فرهنگ رسمی UDS](https://files.alz.washington.edu/documentation/uds3-rdd.pdf) و [فرهنگ ژنتیک](https://files.alz.washington.edu/documentation/rdd-genetic-data.pdf): دامنه‌ها و کدهای متغیرها؛ صفحهٔ دقیق هر قاعده در تنظیمات و فرهنگ منتخب ثبت است.
